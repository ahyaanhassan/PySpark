````python
# Solution 2: Running total of sales per employee

# Write your SQL query here
select
    employee_id,
    employee_name,
    sale_date,
    sales_amount,
    sum(sales_amount) over (
        partition by employee_id
        order by sale_date
        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) as running_total_sales
from employee_sales
order BY employee_id, sale_date;
````

````python
# Solution 3: Employees above department average

# Write your SQL query here
with RankedSales as (
    select
        employee_name,
        region,
        sale_date,
        sales_amount,
        row_number() over (
            partition by region, date_trunc('month', sale_date) #date trunc to get month from Sale Date
            order by sales_amount DESC
        ) as sales_rank
    from employee_sales
)   
select
    employee_id,
    employee_name,
    region,
    sale_date,
    sales_amount
from RankedSales
where sales_rank <= 3
````

````python
# Solution 2: Calculate current balance from transactions

# Write your SQL query here
#use very easy way query so that I understand
#name it current_balance
select
    t.account_id,
    a.account_holder_name,
    a.account_type,
    a.opening_date,
    sum(
        case 
            when t.transaction_type = 'Deposit' then t.amount
            when t.transaction_type = 'Withdrawal' then -t.amount
            when t.transaction_type = 'Transfer' then 
                case 
                    when t.account_id = t.reference_account_id then -t.amount
                    else t.amount
                end
            else 0
        end
    ) as current_balance
from accounts AS a
left join transactions AS t
    on a.account_id = t.account_id
group by
    t.account_id,
    a.account_holder_name,
    a.account_type,
    a.opening_date  
````

**Write your solutions below:**


````python
# Import necessary libraries
import pandas as pd
import numpy as np
from datetime import datetime
import json
import re

# Solution 1: Check completeness function

# Write your code here
def check_completeness(df, threshold=80):
    result = {}
    total_rows = len(df)

    for col_name in df.columns:
        non_null_count = df[col_name].notnull().sum()
        completeness_percent = round((non_null_count / total_rows) * 100, 2)

        # Convert to whole number percentage like the expected output
        result[col_name] = float(int(completeness_percent))

    result["columns_below_threshold"] = [
        col for col, pct in result.items()
        if col != "columns_below_threshold" and pct < threshold
    ]

    return result

df = pd.read_csv("/Users/ahyaan_hassan/Desktop/Pyspark assessment/assessment setup/part_2_python/question_3_python_data_quality/sample_data.csv")
result = check_completeness(df)
print(result)

````

````python
# Solution 2: Check uniqueness function

# Write your code here
def check_uniqueness(df):
    total_rows = len(df)

    duplicate_count = df.duplicated().sum()
    duplicate_percentage = round((duplicate_count / total_rows) * 100, 2)

    return {
        "duplicate_count": int(duplicate_count),
        "duplicate_percentage": duplicate_percentage
    }
result = check_uniqueness(df)
print(result)
````

````python
# Solution 3: Check validity function

# Write your code here
def check_validity(df, rules_dict):
    invalid_results = {}
    
    for col_name, rule in rules_dict.items():
        invalid_indices = []

        # Rule type: numeric range (min, max)
        if isinstance(rule, tuple) and len(rule) == 2:
            min_val, max_val = rule
            for idx, value in df[col_name].items():
                if pd.isnull(value):
                    continue
                if value < min_val or value > max_val:
                    invalid_indices.append(idx)

        # Rule type: email format
        elif rule == "email_format":
            email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
            for idx, value in df[col_name].items():
                if pd.isnull(value):
                    continue
                if not re.match(email_pattern, str(value)):
                    invalid_indices.append(idx)

        # Rule type: date not in future
        elif rule == "not_future_date":
            for idx, value in df[col_name].items():
                if pd.isnull(value):
                    continue
                try:
                    date_val = pd.to_datetime(value)
                    if date_val > pd.Timestamp.today():
                        invalid_indices.append(idx)
                except:
                    invalid_indices.append(idx)

        if invalid_indices:
            invalid_results[col_name] = invalid_indices

    return invalid_results
rules = {
    "age": (0, 120),
    "email": "email_format",
    "last_purchase_date": "not_future_date"
}
validity_result = check_validity(df, rules)
print(validity_result)
````

````python
# Solution 4: Check consistency function

# Write your code here
def check_consistency(df):
    consistency_results = {}

    # Negative or unrealistic ages
    if "age" in df.columns:
        invalid_age_count = df[(df["age"] < 0) | (df["age"] > 120)].shape[0]
        consistency_results["age"] = int(invalid_age_count)

    # Negative purchase amounts
    if "total_purchases" in df.columns:
        invalid_purchase_count = df[df["total_purchases"] < 0].shape[0]
        consistency_results["total_purchases"] = int(invalid_purchase_count)

    # Future dates in last_purchase_date
    if "last_purchase_date" in df.columns:
        future_date_count = 0
        today = pd.Timestamp.today()

        for value in df["last_purchase_date"]:
            if pd.isnull(value):
                continue
            try:
                d = pd.to_datetime(value)
                if d > today:
                    future_date_count += 1
            except:
                future_date_count += 1

        consistency_results["last_purchase_date"] = int(future_date_count)

    return consistency_results
consistency_result = check_consistency(df)
print(consistency_result)
````

````python
# Solution 5: Generate quality report function

# Write your code here
def generate_quality_report(df, rules_dict=None):
    report = {}

    summary = {
        "total_rows": int(len(df)),
        "total_columns": int(len(df.columns)),
        "data_types": {col: str(df[col].dtype) for col in df.columns}
    }

    completeness = check_completeness(df)
    uniqueness = check_uniqueness(df)

    validity = check_validity(df, rules_dict) if rules_dict else {}
    consistency = check_consistency(df)

    report["summary"] = summary
    report["completeness"] = completeness
    report["uniqueness"] = uniqueness
    report["validity"] = validity
    report["consistency"] = consistency

    return report

rules = {
    "age": (0, 120),
    "email": "email_format",
    "last_purchase_date": "not_future_date"
}

quality_report = generate_quality_report(df, rules)
print(json.dumps(quality_report, indent=4))
````
