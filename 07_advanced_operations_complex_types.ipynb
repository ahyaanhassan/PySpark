```python
from pyspark.sql import SparkSession
from pyspark.sql.types import *
# Change 'map' to 'create_map'
from pyspark.sql.functions import col, explode, array, struct, create_map, size, array_contains

# Create SparkSession
spark = SparkSession.builder \
    .appName("Complex Data Types") \
    .master("local[*]") \
    .getOrCreate()

print("SparkSession successfully created!")
```

```python
data = [
    ("Alice", ["Python", "SQL", "Spark"]),
    ("Bob", ["Java", "Scala"]),
    ("Charlie", ["Python", "R", "SQL", "Spark"]),
    ("Diana", ["SQL"])
]

schema = StructType([
    StructField("Name", StringType(), True),
    StructField("Skills", ArrayType(StringType()), True)
])

df_arrays = spark.createDataFrame(data, schema)
df_arrays.show(truncate=False)
df_arrays.printSchema()
```

```python
#access array element by index
df_arrays.withColumn("Firstskill",df_arrays.Name).show(truncate=False)
```

```python
df_has_python_skill = df_arrays.withColumn("HasPythonSkill", array_contains(col("Skills"), "Python"))
df_has_python_skill.show(truncate=False)
```
